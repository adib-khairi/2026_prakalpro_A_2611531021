## 2.1 Dasar Teori

Praktikum Pengenalan Python, VS Code, dan Koneksi ke GitHub membahas dasar penggunaan bahasa pemrograman Python, Visual Studio Code sebagai code editor, serta GitHub sebagai media penyimpanan dan pengelolaan kode program.

1. Python<br>
Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana sehingga mudah dipahami oleh pemula. Python dapat digunakan untuk berbagai kebutuhan, seperti pembuatan aplikasi, pengolahan data, dan pengembangan perangkat lunak.
2. Visual Studio Code<br>
Visual Studio Code (VS Code) merupakan code editor yang digunakan untuk menulis, mengedit, dan menjalankan kode program. VS Code mendukung berbagai bahasa pemrograman, termasuk Python, serta menyediakan fitur seperti terminal, debugging, dan integrasi dengan Git.
3. Git<br>
Git merupakan sistem version control yang digunakan untuk mencatat perubahan pada kode program. Dengan Git, programmer dapat mengelola berbagai versi program dan mengetahui perubahan yang telah dilakukan.
4. GitHub<br>
GitHub merupakan platform berbasis daring yang digunakan untuk menyimpan dan mengelola repository Git. GitHub dapat membantu programmer dalam melakukan pencadangan kode, mengakses proyek dari berbagai perangkat, serta bekerja sama dalam pengembangan program.
5. Repository<br>
Repository merupakan tempat penyimpanan proyek yang berisi kode program dan berbagai berkas pendukung. Repository dapat disimpan secara lokal pada komputer maupun secara daring pada GitHub.
6. Commit dan Push<br>
Commit digunakan untuk menyimpan perubahan yang telah dilakukan pada repository lokal. Push digunakan untuk mengirim perubahan dari repository lokal ke repository yang berada di GitHub.

### 2.2.1 Program Numeric_2611531021

```python
#create a variable with integer value
a_1021 = 100
print("The type of variable having value",a_1021,"is",type(a_1021))

#create variable with float value
b_1021 = 10.2345
print("The type of variable having value",b_1021,"is",type(b_1021))

#create variable with complex value
c_1021 = 100 + 3j
print("The type of variable having value",c_1021,"is",type(c_1021))
```

### 2.2.3 Program Boolean_2611531021

```python
# Buar file dengan nama Boolean_2611531021
# Nama variabel ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean
is_lulus_1021 = True
is_cumlaude_1021 = True

# Menggunakan Boolean
nilai_1021 = 85
batas_lulus_1021 = 75

# Menantukan nilai Boolean dari kondisi
status_kelulusan_1021 = nilai_1021 >= batas_lulus_1021 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai: ",nilai_1021)
print("Apakah lulus?: ",status_kelulusan_1021)
if is_lulus_1021 and is_cumlaude_1021:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

### 2.2.4 Program Input_2611531021

```python
# Buat file dengan nama Input_261151021.py
# Buat Program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit mnimm terakhir contoh: nama_1021
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_1021 = input("Masukkan nilai panjang: ")
lebar_1021 = input("Masukkan nilai lebar: ")
luas_1021 = int(panjang_1021) * int(lebar_1021)
print("Luas = ",luas_1021)
```

### 2.2.5 Program Konstanta_2611531021

```python
# Buat file dengan nama Konstanta_2611531021
# Program ini menggunakan konstanta untuk menghitun luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1021

from typing import Final
PI: Final = 3.14
print("Pi: %f" % (PI))
jari_1021 = float(input("Masukkan nilai jari-jari: "))
luas_1021 = PI * jari_1021
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1021,luas_1021))
```
